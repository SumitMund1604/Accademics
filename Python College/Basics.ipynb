```python
import pandas as pd

weather_data = [
    ['01-01-2017', 32, 6, 20, 'rain'], 
    ['01-02-2017', 35, 7, 30, 'sunny'], 
    ['04-05-2017', 28, 2, 45, 'snow'], 
    ['01-04-2017', 24, 7, 40, 'snow'], 
    ['01-05-2017', 32, 4, 25, 'rainy'], 
    ['01-06-2017', 31, 2, 20, 'sunny']
]

df = pd.DataFrame(weather_data, columns=['day', 'temp', 'windspeed', 'humidity', 'event'])

print(df)
```

```python
from scipy.stats import norm
axes = plt.axes()
axes.set_xlim([-10,10])
axes.set_ylim([0,1.0])
axes.grid()
plt.legend(['pdf','pdf with your val'])
plt.plot(x,norm.pdf(x))
plt.plot(x,norm.pdf(x,1.0,0.5))
plt.show()



```

## 7. Normal Distribution / PDF

```python
from scipy.stats import norm

x = np.arange(-3, 3, 0.0001)

plt.plot(x, norm.pdf(x))
plt.show()
```

## 8. Setting X-axis, Y-axis Limits and Grid

```python
from scipy.stats import norm

x = np.arange(-10, 10, 0.0001)

axes = plt.axes()
axes.set_xlim([-10, 10])
axes.set_ylim([0, 1.0])
axes.grid()

plt.plot(x, norm.pdf(x))
plt.show()
```

## 9. Plotting Two Normal Distributions

```python
from scipy.stats import norm

x = np.arange(-10, 10, 0.0001)

axes = plt.axes()
axes.set_xlim([-10, 10])
axes.set_ylim([0, 1.0])
axes.grid()

plt.plot(x, norm.pdf(x))
plt.plot(x, norm.pdf(x, 1.0, 0.5))

plt.legend(['pdf', 'pdf with your val'])
plt.show()
```

## 11. Bar Chart

A bar chart represents categorical data using rectangular bars.

```python
x = ['A', 'B', 'C', 'D']
y = [10, 20, 15, 25]

plt.bar(x, y)

plt.title('Simple Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

plt.show()
```

## 12. Bar Chart with Color and Width

```python
x = ['A', 'B', 'C', 'D']
y = [10, 20, 15, 25]

plt.bar(x, y, color='green', width=0.5)

plt.title('Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

plt.show()
```

## 13. Horizontal Bar Chart

plt.barh() to create a horizontal bar chart.

```python
x = ['A', 'B', 'C', 'D']
y = [10, 20, 15, 25]

plt.barh(x, y)

plt.title('Horizontal Bar Chart')
plt.xlabel('Values')
plt.ylabel('Categories')

plt.show()
```

## 14. Histogram

A histogram shows the distribution or frequency of numerical data

```python
data = [
    12, 15, 18, 20, 22,
    22, 25, 28, 30, 30,
    32, 35, 35, 38, 40,
    42, 45, 45, 48, 50
]

plt.hist(data)

plt.title('Histogram')
plt.xlabel('Values')
plt.ylabel('Frequency')

plt.show()
```

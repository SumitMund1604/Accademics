## Implementation of lda

```python
import numpy as np
```

```python
s1 = np.array([(4,1),(2,4),(2,3),(3,6),(4,4)])
s2 = np.array([(9,10),(6,8),(9,5),(8,7),(10,8)])
```

```python
print(s1)
print(s2)
```

```python
c1 = s1.mean(axis=0)
c2 = s2.mean(axis=0)
```

```python
print(c1)
print(c2)
```

```python
sw1 = (s1-c1).T @ (s1 - c1)/5
sw2 = (s2-c2).T @ (s2 - c2)/5
sw = sw1 + sw2
```

```python
print(sw1)
```

```python
print(sw2)
```

```python
sw
```

```python
diff = (c1 - c2).reshape(-1,1)
```

```python
diff
```

```python
sb = diff @ diff.T
```

```python
sb
```

```python
matrix = np.linalg.inv(sw) @ sb 
```

```python
matrix
```

```python
eigenvalues, eigenvectors = np.linalg.eig(matrix)
```

```python
eigenvalues
```

```python
eigenvectors
```

```python
best_val = np.argmax(eigenvalues)
```

```python
best_val
```

```python
w = eigenvectors[:,best_val]
```

```python
w
```

```python
lamb = max(np.real(np.linalg.eigvals(matrix)))
```

```python
lamb
```

```python
t = matrix - np.eye(2) * lamb
```

```python
np.linalg.solve(t, np.zeros(t.shape[0]))
```

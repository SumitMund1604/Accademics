
### 3. Initialize the Imputer
* **What it does**: Configures the imputation tool with two rules:
  * `missing_values=np.nan`: Look for empty or blank cells (represented as `NaN` in Python).
  * `strategy="mean"`: Calculate the mathematical average of that column to fill those empty cells.
